## Code – Heart Disease Prediction

### Step 1: Import Libraries
In this step, we import all essential Python libraries required for data manipulation, categorical encoding, model training, evaluation metrics, and tabular display.

````python
"""
Purpose of this code:
----------------------
This script imports the required libraries for building and evaluating a
machine learning model. It sets up tools for data manipulation, preprocessing,
splitting datasets, training a Support Vector Machine (SVM) model,
evaluating accuracy, precision, recall, F1-score, and displaying results in tabular form.
"""

# Import numerical computing library (used for arrays, vectorized math operations)
import numpy as np

# Import data manipulation library (used for DataFrames, file I/O, filtering)
import pandas as pd

# Import pickle (used for serializing and deserializing trained ML models to disk)
import pickle

# Import train_test_split (used to divide dataset into training and testing sets)
from sklearn.model_selection import train_test_split

# Import LabelEncoder (used to convert categorical strings into numeric codes)
from sklearn.preprocessing import LabelEncoder

# Import Support Vector Machine (SVM) from scikit-learn
from sklearn import svm

# Import evaluation metrics for classification
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

# Import PrettyTable (used to display outputs in clean ASCII tables for presentation)
from prettytable import PrettyTable

# Import plotting libraries (used for rendering the confusion matrix heatmap)
import matplotlib.pyplot as plt
import seaborn as sns

print("All required libraries imported successfully!")
````

### Step 2: Load Sample Data
In this step, we load the raw UCI Cleveland Heart Disease dataset from `processed.cleveland.data`. Because this raw file does not contain a header row, we explicitly assign the official 14 attribute column names according to the UCI repository documentation.

````python
"""
Purpose of this section:
-------------------------
This part of the code loads the raw UCI Cleveland Heart Disease dataset
from 'processed.cleveland.data'. Since the raw file has no column headers,
we manually assign the 14 standard medical attribute names, specify that
missing values are encoded as '?', and inspect the dataset shape and first rows.
"""

# Define the standard 14 column names for the UCI Heart Disease dataset
columns = [
    'age', 'sex', 'cp', 'trestbps', 'chol', 'fbs',
    'restecg', 'thalach', 'exang', 'oldpeak', 'slope',
    'ca', 'thal', 'num'
]

# Read the raw dataset, marking '?' as missing values (NaN)
raw_data = pd.read_csv("processed.cleveland.data", names=columns, na_values="?")

print("\nRaw Cleveland Heart Disease Dataset:")
print("====================================\n")
print("Dataset Shape (Rows, Columns):", raw_data.shape)

# Display the first 5 records
print("\nFirst 5 Records:")
print(raw_data.head())
````

### Step 3: Understand and Pre-process Sample Data
In this section, we analyze the structure of the dataset, check for missing values, binarize the clinical target attribute, and map numerical categories into descriptive medical labels.

#### Step 3.1: Understand Sample Data
We inspect the attribute list, count instances, and check for missing values marked with `?`.

````python
"""
Purpose of this section:
-------------------------
This part helps us understand the raw dataset by:
1. Printing all column names (attributes).
2. Printing total instance count.
3. Checking the distribution of missing values across all columns.
"""

print("\nAttributes in Raw Dataset:")
print("==========================\n")
print(raw_data.columns.tolist())

print("\n\nTotal Number of Instances in Raw Data:", len(raw_data))
print("======================================\n")

print("\nMissing Values Count Per Column:")
print("================================\n")
print(raw_data.isna().sum())
````

#### Step 3.2: Pre-process Sample Data
The raw target column `num` contains values from `0` to `4`:
- `0`: Absence of coronary heart disease.
- `1, 2, 3, 4`: Presence of heart disease with varying angiographic severity.

Following standard clinical research protocols, we binarize the target:
- `num == 0` $\rightarrow$ `0` (**No Disease**)
- `num > 0` $\rightarrow$ `1` (**Heart Disease Present**)

We also map `sex` and `cp` into descriptive labels and create `sample-data.csv`.

````python
"""
Purpose of this section:
-------------------------
1. Convert the multi-class target 'num' (0-4) into binary target:
   0 = No heart disease, 1 = Heart disease present.
2. Map numerical codes into clear descriptive categories:
   - sex: 0 -> 'Female', 1 -> 'Male'
   - cp: 1 -> 'Typical Angina', 2 -> 'Atypical Angina', 3 -> 'Non-Anginal Pain', 4 -> 'Asymptomatic'
3. Save the clean sample dataset to 'sample-data.csv'.
"""

# Binarize the target: 0 if num == 0 else 1
raw_data['HeartDisease_Binary'] = (raw_data['num'] > 0).astype(int)

# Dictionaries for descriptive mapping
gender_mapping = {0.0: 'Female', 1.0: 'Male'}
chest_pain_mapping = {
    1.0: 'Typical Angina',
    2.0: 'Atypical Angina',
    3.0: 'Non-Anginal Pain',
    4.0: 'Asymptomatic'
}
target_mapping = {0: 'No', 1: 'Yes'}

# Construct a clean DataFrame with the 4 selected features + target
sample_data = pd.DataFrame({
    'Age': raw_data['age'].astype(int),
    'Gender': raw_data['sex'].map(gender_mapping),
    'ChestPain': raw_data['cp'].map(chest_pain_mapping),
    'MaxHeartRate': raw_data['thalach'].astype(int),
    'HeartDisease': raw_data['HeartDisease_Binary'].map(target_mapping)
})

# Save to sample-data.csv (matching Titanic's sample-data.csv structure)
sample_data.to_csv("sample-data.csv", index=False)
print("Saved clean data to 'sample-data.csv' with shape:", sample_data.shape)

# Display sample records
print("\nSample Data (First 10 Rows):")
print("============================\n")
print(sample_data.head(10))
````

### Step 4: Feature Extraction (Constraint: Exactly 4 Features)

To match the Titanic reference project constraint (which used 4 inputs: `PClass`, `Gender`, `Sibling`, `Embarked`), we select **exactly 4 features**:

1. **`Age`** *(Numerical)*: Age in years.
2. **`Gender`** *(Categorical)*: Biological sex (`Male` / `Female`).
3. **`ChestPain`** *(Categorical)*: Chest pain presentation (`Typical Angina`, `Atypical Angina`, `Non-Anginal Pain`, `Asymptomatic`).
4. **`MaxHeartRate`** *(Numerical)*: Peak heart rate achieved under exercise testing (`thalach`).

#### Justification for Feature Selection:
- **Balanced Feature Mixture:** It provides a balanced combination of **2 continuous numerical attributes** (`Age`, `MaxHeartRate`) and **2 categorical attributes** (`Gender`, `ChestPain`).
- **High Clinical Relevance:** In cardiology, chest pain type is the primary symptomatic diagnostic indicator for myocardial ischemia. Maximum heart rate achieved during exercise stress testing directly measures cardiac reserve. Age and male sex are universally recognized primary demographic risk factors for coronary artery disease.
- **Zero Missing Values:** In the UCI Cleveland dataset, missing values (`?`) only exist in the fluoroscopy feature `ca` (4 missing) and thallium scan `thal` (2 missing). Our 4 selected features have **0 missing values across all 303 patient records**, ensuring a clean, robust training dataset without requiring imputation.

````python
"""
Purpose of this section:
-------------------------
Verifies that our 4 selected features have zero missing values and
displays summary statistics for the selected features.
"""

print("Selected 4 Input Features:")
print("--------------------------")
print("1. Age          (Numerical)")
print("2. Gender       (Categorical: Female, Male)")
print("3. ChestPain    (Categorical: Typical Angina, Atypical Angina, Non-Anginal Pain, Asymptomatic)")
print("4. MaxHeartRate (Numerical)")
print("\nTarget Attribute:")
print("-----------------")
print("HeartDisease    (Binary: No, Yes)")

print("\nMissing Value Verification for Selected Columns:")
print("================================================")
print(sample_data.isna().sum())

print("\nStatistical Summary of Numerical Features:")
print("===========================================")
print(sample_data[['Age', 'MaxHeartRate']].describe())
````

### Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)
Machine learning algorithms (such as Support Vector Machines) require numerical inputs. In this step, we train `LabelEncoder` objects on fixed category sets and transform both output and input attributes into numeric codes, precisely following the Titanic notebook workflow.

#### Step 5.1: Train the Label Encoder
We define fixed category lists for each categorical column and fit separate `LabelEncoder` objects.

````python
"""
Purpose of this section:
-------------------------
This part of the code sets up and trains Label Encoders.
Label Encoding converts categorical values into numeric codes:
- Gender: 'Female' -> 0, 'Male' -> 1
- ChestPain: 'Asymptomatic' -> 0, 'Atypical Angina' -> 1, 'Non-Anginal Pain' -> 2, 'Typical Angina' -> 3
- HeartDisease: 'No' -> 0, 'Yes' -> 1
"""

# Define the category sets for each categorical attribute
gender_categories = pd.DataFrame({"Gender": ["Female", "Male"]})
chest_pain_categories = pd.DataFrame({"ChestPain": ["Asymptomatic", "Atypical Angina", "Non-Anginal Pain", "Typical Angina"]})
heart_disease_categories = pd.DataFrame({"HeartDisease": ["No", "Yes"]})

# Create LabelEncoder instances
gender_label_encoder = LabelEncoder()
chest_pain_label_encoder = LabelEncoder()
heart_disease_label_encoder = LabelEncoder()

# Train (fit) each encoder on its defined categories
gender_label_encoder.fit(np.ravel(gender_categories))
chest_pain_label_encoder.fit(np.ravel(chest_pain_categories))
heart_disease_label_encoder.fit(np.ravel(heart_disease_categories))

print("Label Encoders trained successfully!")
print("Gender mapping:     ", dict(zip(gender_label_encoder.classes_, gender_label_encoder.transform(gender_label_encoder.classes_))))
print("Chest Pain mapping: ", dict(zip(chest_pain_label_encoder.classes_, chest_pain_label_encoder.transform(chest_pain_label_encoder.classes_))))
print("Heart Disease mapping:", dict(zip(heart_disease_label_encoder.classes_, heart_disease_label_encoder.transform(heart_disease_label_encoder.classes_))))
````

#### Step 5.2: Label Encode the Output
We encode the target attribute `HeartDisease` ('No' $\rightarrow$ 0, 'Yes' $\rightarrow$ 1) and save `sample-data-encoded-output.csv`.

````python
"""
Purpose of this section:
-------------------------
Applies Label Encoding to the target variable ("HeartDisease").
Categorical values ("No", "Yes") are converted into numerical codes (0 and 1).
"""

# Create a copy of the dataset
sample_data_encoded_output = sample_data.copy()

# Transform 'HeartDisease' using the fitted encoder
sample_data["encoded_heart_disease"] = heart_disease_label_encoder.transform(sample_data["HeartDisease"])

print("\nHeartDisease Attribute After Label Encoding:")
print("============================================\n")
print(sample_data[["HeartDisease", "encoded_heart_disease"]].head(10))

# Replace the categorical output with the encoded numeric output
sample_data_encoded_output['HeartDisease'] = sample_data['encoded_heart_disease']

# Save to CSV
sample_data_encoded_output.to_csv("sample-data-encoded-output.csv", index=False)
print("\nSaved: sample-data-encoded-output.csv")
````

#### Step 5.3: Label Encode the Input
We encode the categorical input attributes (`Gender` and `ChestPain`) into numeric codes and save `sample-data-encoded.csv`.

````python
"""
Purpose of this section:
-------------------------
Applies Label Encoding to categorical input features ('Gender' and 'ChestPain').
Numerical features ('Age' and 'MaxHeartRate') remain numeric as they already are.
"""

# Create a copy of the encoded-output dataframe
sample_data_encoded = sample_data_encoded_output.copy()

# Transform categorical input attributes
sample_data_encoded['Gender'] = gender_label_encoder.transform(sample_data_encoded['Gender'])
sample_data_encoded['ChestPain'] = chest_pain_label_encoder.transform(sample_data_encoded['ChestPain'])

print("\nSample Data After Complete Label Encoding (First 10 Rows):")
print("===========================================================\n")
print(sample_data_encoded.head(10))

# Save the completely encoded dataset to CSV
sample_data_encoded.to_csv("sample-data-encoded.csv", index=False)
print("\nSaved: sample-data-encoded.csv")
````

### Step 6: Execute the Training Phase
In this step, we split our encoded dataset into training (80%) and testing (20%) partitions, separate feature vectors from labels, train a Support Vector Classifier (SVC), and save the trained model.

#### Step 6.1: Splitting Sample Data into Training Data and Testing Data
We perform an 80/20 train/test split. Setting `random_state=42` and `shuffle=True` guarantees a balanced, reproducible distribution of patient records across both sets.

````python
"""
Purpose of this section:
-------------------------
Splits the encoded dataset into Training Data (80%) and Testing Data (20%).
- Training data is used by the SVC algorithm to learn the decision boundary.
- Testing data is held out to evaluate how well the model generalizes to unseen patients.
"""

training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,
    random_state=42,
    shuffle=True
)

# Save the training and testing datasets to CSV files
training_data_encoded.to_csv("training-data-encoded.csv", index=False)
testing_data_encoded.to_csv("testing-data-encoded.csv", index=False)

print("Training Data shape: ", training_data_encoded.shape)
print("Testing Data shape:  ", testing_data_encoded.shape)
print("Saved: training-data-encoded.csv and testing-data-encoded.csv")
````

#### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data
We extract the input features (`Age`, `Gender`, `ChestPain`, `MaxHeartRate`) as `input_vector_train` and the target (`HeartDisease`) as `output_label_train`.

````python
"""
Purpose of this section:
-------------------------
Separates the input features (X) from the target label (y) in the training dataset.
"""

# Extract input features (all columns except the last one)
input_vector_train = training_data_encoded.iloc[:, :-1]

# Extract target label (the last column)
output_label_train = training_data_encoded.iloc[:, -1]

print("Input vectors (train) shape: ", input_vector_train.shape)
print("Output labels (train) shape: ", output_label_train.shape)
print("\nFirst 5 Training Feature Vectors:")
print(input_vector_train.head())
````

#### Step 6.3: Train the Support Vector Classifier
We initialize and train a Support Vector Classifier (`SVC`). We use a `linear` kernel (which provides clear linear separation in the 4-dimensional feature space) and enable `probability=True` so the model can generate confidence scores in our Streamlit web application.

````python
"""
Purpose of this section:
-------------------------
Trains a Support Vector Classifier (SVC) model from scikit-learn.
The linear kernel finds the optimal maximum-margin hyperplane separating
patients with heart disease from healthy individuals.
"""

print("\nTraining Support Vector Classifier on Training Data...")
print("=====================================================\n")

# Initialize SVC with linear kernel and probability calibration
svc_model = svm.SVC(kernel='linear', probability=True, random_state=42)

# Train the model using features and target labels
svc_model.fit(input_vector_train, np.ravel(output_label_train))

print("Model Training Completed!")
print(svc_model)
````

#### Step 6.4: Save the Trained Model
We save the trained SVC model to `heart_model.pkl` using Python's `pickle` library, allowing it to be loaded directly by our Streamlit web app without retraining.

````python
"""
Purpose of this section:
-------------------------
Saves the trained model and associated encoders to disk using pickle.
"""

# Save the trained model to 'heart_model.pkl'
with open('heart_model.pkl', 'wb') as f:
    pickle.dump(svc_model, f)

print("Saved trained model to: heart_model.pkl")
````

### Step 7: Execute the Testing Phase
In this section, we evaluate the trained model on the held-out testing partition (61 patients) to rigorously assess accuracy, precision, recall, F1-score, and the confusion matrix.

#### Step 7.1: Splitting Input Vectors and Outputs / Labels of Testing Data
We separate the testing features and actual ground-truth labels.

````python
"""
Purpose of this section:
-------------------------
Separates features and ground-truth labels for the held-out testing dataset.
"""

input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

print("Input vectors (test) shape: ", input_vector_test.shape)
print("Output labels (test) shape: ", output_label_test.shape)
````

#### Step 7.2: Load the Saved Model
We load `heart_model.pkl` back from disk to verify that model serialization works properly.

````python
"""
Purpose of this section:
-------------------------
Loads the previously saved Support Vector Classifier from 'heart_model.pkl'.
"""

model = pickle.load(open('heart_model.pkl', 'rb'))
print("Model loaded successfully from heart_model.pkl!")
````

#### Step 7.3: Evaluate the Machine Learning Model
We generate predictions on the test instances and save the comparison between actual and predicted values to `model-predictions.csv`.

````python
"""
Purpose of this section:
-------------------------
Generates predictions on test data and stores them alongside actual labels.
"""

model_predictions = model.predict(input_vector_test)

# Attach predictions to test dataframe
test_results = testing_data_encoded.copy()
test_results["Predictions"] = model_predictions

# Save predictions to CSV
test_results.to_csv("model-predictions.csv", index=False)
print("Saved predictions to: model-predictions.csv")

print("\nSample Predictions vs Actual Labels (First 10 Patients):")
print("========================================================\n")
print(test_results.head(10))
````

#### Step 7.4: Calculate the Accuracy Score
We compute overall test accuracy: the percentage of test instances where the model's prediction matched the actual clinical ground truth.

````python
"""
Purpose of this section:
-------------------------
Computes and prints test accuracy score.
"""

accuracy = accuracy_score(test_results["HeartDisease"], test_results["Predictions"])

print("\nAccuracy Score on Held-Out Test Set:")
print("====================================\n")
print(f"Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")
````

#### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix Heatmap)
In medical diagnosis, accuracy alone is insufficient:
- **Precision:** Of all patients predicted to have heart disease, what percentage actually had it?
- **Recall (Sensitivity):** Of all patients who actually had heart disease, what percentage did the model correctly identify?
- **F1-Score:** The harmonic mean of precision and recall.
- **Confusion Matrix:** True Negatives, False Positives, False Negatives, and True Positives visualized as a Seaborn heatmap.

````python
"""
Purpose of this section:
-------------------------
Generates comprehensive clinical evaluation metrics:
Precision, Recall, F1-Score, Classification Report, and Confusion Matrix Heatmap.
"""

y_true = test_results["HeartDisease"]
y_pred = test_results["Predictions"]

print("Classification Report:")
print("======================\n")
print(classification_report(
    y_true,
    y_pred,
    target_names=["No Heart Disease (0)", "Heart Disease Present (1)"]
))

# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
cm_df = pd.DataFrame(
    cm,
    index=["Actual: No Disease", "Actual: Heart Disease"],
    columns=["Predicted: No Disease", "Predicted: Heart Disease"]
)

print("\nConfusion Matrix Table:")
print("=======================\n")
print(cm_df)

# Plot Confusion Matrix Heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Reds',
            xticklabels=['No Disease (0)', 'Disease (1)'],
            yticklabels=['No Disease (0)', 'Disease (1)'])
plt.xlabel('Predicted Label', fontsize=11)
plt.ylabel('Actual Label', fontsize=11)
plt.title('Confusion Matrix - SVC Heart Disease Predictor', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
````

### Step 8: Execute the Application Phase
In this phase, we simulate an unseen patient consultation: taking user input, constructing the feature vector, applying label encoding, querying the saved model, and formatting the output in an ASCII PrettyTable with probability estimates.

#### Step 8.1: Take Input from User
We define attributes for a sample patient (e.g. 55-year-old male presenting with Asymptomatic chest pain and max heart rate of 140 bpm).

````python
"""
Purpose of this section:
-------------------------
Takes patient clinical inputs. For non-interactive batch runs, sample patient
values are provided below. In interactive environments, input() can be used.
"""

# Sample patient inputs
age_input = 55
gender_input = "Male"
chest_pain_input = "Asymptomatic"
max_heart_rate_input = 140

print("Patient Attributes Entered:")
print("--------------------------")
print(f"Age:            {age_input}")
print(f"Gender:         {gender_input}")
print(f"Chest Pain:     {chest_pain_input}")
print(f"Max Heart Rate: {max_heart_rate_input} bpm")
````

#### Step 8.2: Convert User Input into Feature Vector
We structure user input into a single-row pandas DataFrame matching the schema of our training data.

````python
"""
Purpose of this section:
-------------------------
Formats raw user input into a structured DataFrame feature vector.
"""

user_input = pd.DataFrame({
    'Age': [age_input],
    'Gender': [gender_input],
    'ChestPain': [chest_pain_input],
    'MaxHeartRate': [max_heart_rate_input]
})

print("\nUser Input Feature Vector:")
print("==========================\n")
print(user_input)
````

#### Step 8.3: Label Encoding of Feature Vector
We encode categorical attributes using the pre-fitted LabelEncoders.

````python
"""
Purpose of this section:
-------------------------
Applies pre-fitted LabelEncoders to the user input feature vector.
"""

unseen_data_features = user_input.copy()
unseen_data_features['Gender'] = gender_label_encoder.transform(user_input['Gender'])
unseen_data_features['ChestPain'] = chest_pain_label_encoder.transform(user_input['ChestPain'])

print("\nUser Input Encoded Feature Vector:")
print("==================================\n")
print(unseen_data_features)
````

#### Step 8.4: Load the Saved Model
We load `heart_model.pkl` for inference.

````python
"""
Purpose of this section:
-------------------------
Loads the saved trained model from heart_model.pkl for inference.
"""

application_model = pickle.load(open('heart_model.pkl', 'rb'))
print("Model loaded successfully for Application Phase!")
````

#### Step 8.5: Model Prediction
We compute the binary prediction and decision probability, formatting the result in a PrettyTable.

````python
"""
Purpose of this section:
-------------------------
Applies the trained SVC model to the encoded feature vector and displays
the prediction and confidence in a clean PrettyTable.
"""

predicted_code = application_model.predict(unseen_data_features)[0]
probabilities = application_model.predict_proba(unseen_data_features)[0]

if predicted_code == 1:
    prediction_text = "HEART DISEASE LIKELY"
    risk_level = "Elevated Risk"
else:
    prediction_text = "HEART DISEASE UNLIKELY"
    risk_level = "Low Risk"

disease_probability = probabilities[1] * 100

pretty_table = PrettyTable()
pretty_table.field_names = ["Patient Parameter", "Value"]
pretty_table.add_row(["Age", age_input])
pretty_table.add_row(["Gender", gender_input])
pretty_table.add_row(["Chest Pain Type", chest_pain_input])
pretty_table.add_row(["Max Heart Rate Achieved", f"{max_heart_rate_input} bpm"])
pretty_table.add_row(["Diagnostic Prediction", prediction_text])
pretty_table.add_row(["Disease Probability", f"{disease_probability:.2f}%"])
pretty_table.add_row(["Clinical Assessment", risk_level])

print("\n\n================ DIAGNOSTIC REPORT ================")
print(pretty_table)
````
